```python
columns = ["Age", "Fare", "SibSp", "Parch", "Pclass"]
```

```python
print(df[columns].mode().iloc[0])
```

```python
df[columns].std()
```

```python
percentiles = df[columns].quantile([0.25, 0.50, 0.75])
print(percentiles)
```

```python
print(df[columns].skew())
```

```python
summary = pd.DataFrame({
    "Mean": df[columns].mean(),
    "Median": df[columns].median(),
    "Mode": df[columns].mode().iloc[0],
    "Std Dev": df[columns].std(),
    "25th Percentile": df[columns].quantile(0.25),
    "50th Percentile": df[columns].quantile(0.50),
    "75th Percentile": df[columns].quantile(0.75),
    "Skewness": df[columns].skew()
})

print(summary)
```

```python
import os

os.makedirs("Excel", exist_ok=True)
```

```python
summary.to_excel(
    "Excel/Task9_Descriptive_Statistics.xlsx",
    index=False
)

print("Excel file created successfully!")
```
